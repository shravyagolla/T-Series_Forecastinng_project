```python
from delta.tables import DeltaTable

df_telemetry = spark.table("workspace.default.bronze_telemetry")

silver_telemetry = (
    df_telemetry
    .dropDuplicates(["datetime", "machineID"])
    .filter(
        "datetime IS NOT NULL AND machineID IS NOT NULL"
    )
)

target_table = "workspace.default.silver_telemetry"

if not spark.catalog.tableExists(target_table):

    silver_telemetry.write.format("delta").saveAsTable(target_table)

else:

    target = DeltaTable.forName(spark, target_table)

    target.alias("t").merge(
        silver_telemetry.alias("s"),
        """
        t.datetime = s.datetime
        AND t.machineID = s.machineID
        """
    ).whenMatchedUpdateAll() \
     .whenNotMatchedInsertAll() \
     .execute()

print("Silver telemetry table created/updated")
silver_telemetry.printSchema()
```
